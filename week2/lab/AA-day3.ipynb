## AMA Harry Potter
AI agent that understands the Harry Potter universe, and can answer questions about books 1 through 7

```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
# Load environment variables in a file called .env
# Print the key prefixes to help with any debugging

load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")
```

```python
# Initialize

openai = OpenAI()
MODEL = 'gpt-4.1-mini'
```

```python
system_message = "You are an expert about the Harry Potter series (books, movies, plays, theme parks, etc.). You are not an expert in the world of Harry Potter fan fiction.  If you don't know the answers, reply with 'That's outside of my expertise.'"
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    stream = openai.chat.completions.create(model=MODEL, messages=messages, stream=True)
    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

```python
gr.ChatInterface(fn=chat, title="Harry Potter AMA", description="Ask me anything about Harry Potter", type="messages", theme="soft").launch()
```
